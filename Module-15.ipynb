2D Line from Pandas DF

```python
import pandas as pd
df = pd.read_csv('enrollment_data.csv')
df
```

```python
plt.xlabel('Year')
plt.ylabel('Number of Students')
plt.title('Enrollment Data')
plt.plot(df['Year'], df['Programming'])
plt.plot(df['Year'], df['Digital Marketing'])
```

```python
#adding lavels using plt.legend() function
plt.plot(df['Year'], df['Programming'], label='Programming')
plt.plot(df['Year'], df['Digital Marketing'], label='Digital Marketing')
plt.legend(loc = 'lower right')
```

```python
# Adding some visualization to the plot using plt.style.use() function
plt.xlabel('Years')
plt.ylabel('Count of Enrolment')
plt.title('Student Enrolment over 8 Years')

plt.plot(df['Year'], df['Programming'], label='Programming', color ='Red', linewidth = 2, linestyle = 'dashed', marker = 'o')
plt.plot(df['Year'], df['Digital Marketing'], label='Digital Marketing', color = 'Black',linewidth = 2, linestyle = 'dashed', marker = 'o')
plt.legend()
```

Bar Chart

```python
data = {
    'Student' : ['A', 'B', 'C', 'D', 'E'],
    'Gender' : ['Male', 'Female', 'Male', 'Female', 'Male'],
    'Study_Hour' : [2, 5, 3, 4, 6],
    'Attendance' : [80, 90, 70, 85, 95],
    'Grade' : ['A', 'B', 'C', 'B', 'A']
}

df = pd.DataFrame(data)
```

```python
df
```

```python
# Gender wise

gender_group = df.groupby('Gender').size()

gender_group.index
gender_group.values
```

```python
plt.bar(gender_group.index, gender_group.values, color = ['orange', 'red'],edgecolor = 'black')
plt.xlabel('Gender')
plt.ylabel('Count of Students')
```

```python
# Bar chartbon real data
df = pd.read_csv('student_data.csv')
df
```

```python
status = df.groupby('CompletionStatus').size()

plt.bar(status.index, status.values, color = ['orange', 'red'],edgecolor = 'black')
plt.xlabel('Gender')
plt.ylabel('Count of Students')
plt.title('Completion Status of Students')
```

```python
status = df.groupby('Location').size()
plt.bar(status.index, status.values, color = ['orange', 'red'],edgecolor = 'black')

plt.xlabel('Location')
plt.ylabel('Count of Students')
plt.title('Completion Location of Students')
```

Pie Chart

```python
status = df.groupby('CompletionStatus').size()
plt.pie(status, labels = status.index, autopct='%1.1f%%')
plt.show()
```
